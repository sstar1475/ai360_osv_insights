```python
from db.OSVDataClient import OSVDataClient
```

```python
df = OSVDataClient()
```

```python
df_report = await df.get_detailed_report()
```

```python
df_report
```

```python
ecosystem_names = ['PyPI', 'Go', 'OSS-Fuzz', 'Maven', 'npm'] 
```

```python
ECOSYSTEM = {}
for name in ecosystem_names:
    ECOSYSTEM.update({name: df_report[df_report['package_ecosystem'] == name]})
```

```python
dct = {}
ans = 0
for name in ecosystem_names:    
    for x in ECOSYSTEM[name]['vulnerability_severity_score'].values:
        ans += 1
        if not isinstance(x, str):
            continue
        if x[:8] not in dct:
            dct.update({x[:8] : 1})
dct, ans
```

```python
import json
from cvss import CVSS2, CVSS3, CVSS4

def extract_cvss_severity(severity):
    if not isinstance(severity, str):
        return 0
    if 'CVSS:4.0' in severity:
        c = CVSS4(severity)
        return c.scores()[0]
    elif 'CVSS:3.' in severity:
        c = CVSS3(severity)
        return c.scores()[0]
    else:
        c = CVSS2(severity)
        return c.scores()[0]
    return 0
```

```python
df_report['vulnerability_severity_score'] = df_report['vulnerability_severity_score'].apply(extract_cvss_severity) 
```
